## Context Relevancy

```python
from dotenv import load_dotenv
from datasets import Dataset 
import os
from ragas import evaluate
from ragas.metrics import context_relevancy


# 加载环境变量
load_dotenv()

# 创建数据样本
data_samples = {
    'question': ['法国的首都是哪里？']*2, 
    'contexts' : [['法国位于西欧，包括中世纪城市、高山村庄和地中海海滩。其首都巴黎以其时装屋、包括卢浮宫在内的古典艺术博物馆和埃菲尔铁塔等古迹而闻名。'], 
    ['法国位于西欧，包括中世纪城市、高山村庄和地中海海滩。其首都巴黎以其时装屋、包括卢浮宫在内的古典艺术博物馆和埃菲尔铁塔等古迹而闻名。该国还以其葡萄酒和精致的美食而闻名。拉斯科（Lascaux）的古代洞穴壁画，里昂的罗马剧院和广阔的凡尔赛宫（Palace of Versailles）证明了其丰富的历史。']],
}

# 从字典中创建Dataset对象
dataset = Dataset.from_dict(data_samples)
print(dataset)

# 评估
score = evaluate(dataset, metrics=[context_relevancy])
print(score)

score.to_pandas()
```

## Faithfulness

```python
from dotenv import load_dotenv
from datasets import Dataset 
import os
from ragas import evaluate
from ragas.metrics import faithfulness


# 加载环境变量
load_dotenv()

# 创建数据样本
data_samples = {
    'question': ['爱因斯坦出生于何时何地？']*2,
    'contexts' : [['阿尔伯特·爱因斯坦（生于 1879 年 3 月 14 日）是一位出生于德国的理论物理学家，被广泛认为是有史以来最伟大、最有影响力的科学家之一']]*2,
    'answer': ['爱因斯坦于1879年3月14日出生于德国。', '爱因斯坦于 1879 年 3 月 20 日出生于德国。']
}

# 从字典中创建Dataset对象
dataset = Dataset.from_dict(data_samples)
print(dataset)

# 评估
score = evaluate(dataset, metrics=[faithfulness])
print(score)

score.to_pandas()
```

## Answer Relevance

```python
from dotenv import load_dotenv
from datasets import Dataset 
import os
from ragas import evaluate
from ragas.metrics import answer_relevancy


# 加载环境变量
load_dotenv()

# 创建数据样本
data_samples = {
    'question': ['法国在哪里，它的首都是什么？']*2,
    'contexts' : [['']]*2,
    'answer': ['法国在西欧。', '法国在西欧，巴黎是其首都。']
}

# 从字典中创建Dataset对象
dataset = Dataset.from_dict(data_samples)
print(dataset)

# 评估
score = evaluate(dataset, metrics=[answer_relevancy])
print(score)

score.to_pandas()
```
