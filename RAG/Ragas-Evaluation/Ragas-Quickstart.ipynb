## 整体

```python
from dotenv import load_dotenv
from datasets import Dataset 
import os
from ragas import evaluate
from ragas.metrics import faithfulness, answer_correctness


# 加载环境变量
load_dotenv()

# 创建数据样本
data_samples = {
    'question': ['第一次超级碗是什么时候？', '谁赢得了最多的超级碗？'],
    'answer': ['第一次超级碗是在1967年1月15日举行的', '赢得最多超级碗的是新英格兰爱国者队'],
    'contexts' : [['第一次AFL-NFL世界冠军赛是一场于1967年1月15日在洛杉矶纪念体育场举行的美式足球比赛，'], 
    ['绿湾包装工队...威斯康星州绿湾市。', '包装工队竞争...足球会议']],
    'ground_truth': ['第一次超级碗是在1967年1月15日举行的', '新英格兰爱国者队已经赢得了创纪录的六次超级碗']
}

# 从字典中创建Dataset对象
dataset = Dataset.from_dict(data_samples)

# 使用evaluate函数评估dataset，采用faithfulness和answer_correctness作为度量标准
score = evaluate(dataset, metrics=[faithfulness,answer_correctness])

# 将评估结果转换为Pandas DataFrame格式
score.to_pandas()
```
