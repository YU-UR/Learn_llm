## 分步实现

### 创建文档分割器

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

# 这个文本拆分器用于创建子文档
child_splitter = RecursiveCharacterTextSplitter(chunk_size=200, chunk_overlap=0, add_start_index=True)

# 这个文本拆分器用于创建父文档
parent_splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=100, add_start_index=True)
```

#### 分割文档

```python
parent_docs = parent_splitter.split_documents(docs)
print(len(parent_docs))
parent_docs
```

```python
import uuid

parent_doc_ids = [str(uuid.uuid4()) for _ in parent_docs]
parent_doc_ids
```

```python
all_sub_docs = []
full_docs = []
for i, doc in enumerate(parent_docs):
    _id = parent_doc_ids[i]
    sub_docs = child_splitter.split_documents([doc])
    for _doc in sub_docs:
        _doc.metadata["doc_id"] = _id
    all_sub_docs.extend(sub_docs)
    full_docs.append((_id, doc))
```

```python
from langchain_community.vectorstores import Qdrant
from qdrant_client import QdrantClient

collection_name = "full_documents"

qdrant = Qdrant(
    client=QdrantClient(url="http://localhost:6333"), 
    collection_name=collection_name, 
    embeddings=embeddings
)
```

```python
from qdrant_client.http.models import Distance, VectorParams

# 向量维度
size = 1536

# 若集合已存在，删除该集合
if qdrant.client.collection_exists(collection_name):
    print(f"集合已存在 | collection_name: {collection_name}")
    qdrant.client.delete_collection(collection_name)
    print(f"集合已删除 | collection_name: {collection_name}")
    
# 创建新集合
if qdrant.client.create_collection(
    collection_name=collection_name,
    vectors_config=VectorParams(
                size=size, distance=Distance.COSINE)
):
    print(f"创建集合成功 | collection_name: {collection_name}")
else:
    print(f"创建集合失败 | collection_name: {collection_name}")
```

```python
query = "京东物流营收如何？"
```

```python
# 搜索与 query 相似度较高的子文档
retrieved_sub_docs = qdrant.similarity_search(query)

print(len(retrieved_sub_docs))

for doc in retrieved_sub_docs:
    print(doc)
    print()

# print(retrieved_sub_docs[0].page_content)
```

```python
# 返回父文档ID
retrieved_parent_ids = []
for d in retrieved_sub_docs:
    if "doc_id" in d.metadata and d.metadata["doc_id"] not in retrieved_parent_ids:
        retrieved_parent_ids.append(d.metadata["doc_id"])
retrieved_parent_ids
```

```python
retrieved_parent_docs = parent_store.mget(retrieved_parent_ids)
print(len(retrieved_parent_docs))

retrieved_parent_docs
```

## 使用ParentDocumentRetriever实现

### 创建文档分割器

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
# 这个文本拆分器用于创建子文档
# 应该创建比父文档更小的文档
child_splitter = RecursiveCharacterTextSplitter(chunk_size=200, chunk_overlap=0, add_start_index=True)

# 这个文本拆分器用于创建父文档
parent_splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=100, add_start_index=True)
```

### 创建父文档存储

```python
from langchain.storage import InMemoryStore
```

```python
# 父文档的存储层
store = InMemoryStore()
```

### 创建子文档存储 —— 向量存储

```python
from langchain_openai import OpenAIEmbeddings

embeddings = OpenAIEmbeddings()
```

```python
from langchain_community.vectorstores import Qdrant
from qdrant_client import QdrantClient

collection_name = "full_documents_1"

qdrant = Qdrant(
    client=QdrantClient(url="http://localhost:6333"), 
    collection_name=collection_name, 
    embeddings=embeddings
)
```

```python
from qdrant_client.http.models import Distance, VectorParams

# 向量维度
size = 1536

# 若集合已存在，删除该集合
if qdrant.client.collection_exists(collection_name):
    print(f"集合已存在 | collection_name: {collection_name}")
    qdrant.client.delete_collection(collection_name)
    print(f"集合已删除 | collection_name: {collection_name}")
    
# 创建新集合
if qdrant.client.create_collection(
    collection_name=collection_name,
    vectors_config=VectorParams(
                size=size, distance=Distance.COSINE)
):
    print(f"创建集合成功 | collection_name: {collection_name}")
else:
    print(f"创建集合失败 | collection_name: {collection_name}")
```

### 创建检索器

```python
from langchain.retrievers import ParentDocumentRetriever
```

```python
# 初始化检索器
retriever = ParentDocumentRetriever(
    vectorstore=qdrant,
    docstore=store,
    child_splitter=child_splitter,
    parent_splitter=parent_splitter,
)
```

### 存储文档

```python
store
```

```python
# 查看store中存储的父文档的ID
list(store.yield_keys())
```

```python
docs
```

```python
# 将文档添加到文档存储和向量存储中
# ids:文档的ID列表。如果未提供，则生成UUID。
retriever.add_documents(docs, ids=None)
```

```python
# 查看store中存储的父文档的ID
list(store.yield_keys())
```

### 检索

检索器返回的是父文档（大chunk），即子文档（小chunk）所属的父文档（大chunk）

```python
query = "京东物流营收情况如何？"
```

```python
retrieved_docs = retriever.invoke(query)

print(len(retrieved_docs))
print(retrieved_docs)
```
