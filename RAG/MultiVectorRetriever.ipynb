## Summary

### 创建sub_docs —— summary_docs

```python
import uuid

from langchain_core.documents import Document
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import ChatPromptTemplate
from langchain_openai import ChatOpenAI
```

```python
chain = (
    {"doc": lambda x: x.page_content}
    | ChatPromptTemplate.from_template("总结以下文档：\n\n{doc}")
    | ChatOpenAI(max_retries=0)
    | StrOutputParser()
)
```

```python
docs[0]
```

```python
chain.invoke(docs[0])
```

```python
summaries = chain.batch(docs, {"max_concurrency": 5})
```

```python
summaries
```

```python
docs
```

```python
doc_ids = [str(uuid.uuid4()) for _ in docs]
doc_ids
```

```python
summary_docs = [
    Document(page_content=s, metadata={"doc_id": doc_ids[i]})
    for i, s in enumerate(summaries)
]
```

```python
summary_docs
```

### 创建检索器

```python
# 用于索引子文档的 vectorstore
vectorstore = Chroma(collection_name="summaries", embedding_function=OpenAIEmbeddings())

# 父文档的存储层
store = InMemoryStore()

# 检索器（开始时为空）
retriever = MultiVectorRetriever(
    vectorstore=vectorstore,
    docstore=store,
)
```

### 存储文档

```python
retriever.vectorstore.add_documents(summary_docs)
```

```python
retriever.docstore.mset(list(zip(doc_ids, docs)))
```

```python
list(zip(doc_ids, docs))
```

### 检索

#### 测试只使用向量数据库的检索

```python
query = "介绍下腾讯的大模型"
```

```python
sub_docs = vectorstore.similarity_search(query)
```

```python
len(sub_docs)
```

```python
sub_docs
```

#### 测试使用检索器的检索

```python
retrieved_docs = retriever.invoke(query)
```

```python
len(retrieved_docs)
```

```python
len(retrieved_docs[0].page_content)
```

```python
retrieved_docs
```

## Hypothetical Queries

### 创建sub_docs —— question_docs

```python
functions = [
    {
        "name": "hypothetical_questions",
        "description": "生成假设性问题",
        "parameters": {
            "type": "object",
            "properties": {
                "questions": {
                    "type": "array",
                    "items": {"type": "string"},
                },
            },
            "required": ["questions"],
        },
    }
]
```

```python
from langchain.output_parsers.openai_functions import JsonKeyOutputFunctionsParser

chain = (
    {"doc": lambda x: x.page_content}
    # 只要求生成3个假设性问题，但这个数量是可以调整的
    | ChatPromptTemplate.from_template(
        "生成一个列表，包含正好3个可以用以下文档回答的假设性问题:\n\n{doc}"
    )
    | ChatOpenAI(max_retries=0, model="gpt-4").bind(
        functions=functions, function_call={"name": "hypothetical_questions"}
    )
    | JsonKeyOutputFunctionsParser(key_name="questions")
)
```

```python
docs[0]
```

```python
chain.invoke(docs[0])
```

```python
hypothetical_questions = chain.batch(docs, {"max_concurrency": 5})
```

```python
hypothetical_questions
```

```python
# 为每个文档生成唯一标识符
doc_ids = [str(uuid.uuid4()) for _ in docs]
```

```python
doc_ids
```

```python
question_docs = []
for i, question_list in enumerate(hypothetical_questions):
    question_docs.extend(
        [Document(page_content=s, metadata={"doc_id": doc_ids[i]}) for s in question_list]
    )
question_docs
```

### 创建检索器

```python
# 用于索引子块的向量存储
vectorstore = Chroma(
    collection_name="hypo-questions", embedding_function=OpenAIEmbeddings()
)

# 父文档的存储层
store = InMemoryStore()


# 检索器（初始为空）
retriever = MultiVectorRetriever(
    vectorstore=vectorstore,
    docstore=store,
)
```

### 存储文档

```python
retriever.vectorstore.add_documents(question_docs)
```

```python
retriever.docstore.mset(list(zip(doc_ids, docs)))
```

```python
list(zip(doc_ids, docs))
```
